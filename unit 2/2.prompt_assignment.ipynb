def reverse_string(s: str) -> str:
    """
    Reverses a given string using recursion.

    This function takes a string and returns a new string with its characters
    in reverse order. It achieves this by recursively building the reversed
    string from the last character to the first.

    Constraints:
    - You MUST use recursion.
    - You are NOT allowed to use string slicing (e.g., [::-1]).
      To strictly adhere to this, the implementation avoids even common
      recursive string slicing patterns like `s[1:]` or `s[:-1]` by
      using an index-based helper function. This prevents the creation
      of intermediate string objects for subproblems, though string
      concatenation still creates new strings.

    Args:
        s (str): The input string to be reversed.

    Returns:
        str: The reversed string.

    Raises:
        TypeError: If the input `s` is not a string.

    Time Complexity:
        O(N^2), where N is the length of the input string `s`.
        This complexity arises primarily from string concatenation in Python.
        When `char + sub_result` is performed, a new string object is created,
        and characters from `sub_result` are copied into it along with `char`.
        In the worst case, this involves copying strings of lengths 1, 2, ..., N.
        The sum of these copy operations is 1 + 2 + ... + N, which is O(N^2).

    Space Complexity:
        O(N), where N is the length of the input string `s`.
        This accounts for two main factors:
        1.  Recursion Call Stack: The depth of the recursion is N, as each
            recursive call processes one character. Each stack frame consumes
            a constant amount of space.
        2.  Intermediate String Objects: Although string concatenation leads
            to O(N^2) time, the maximum space used for intermediate string
            objects at any given point in time is O(N) (e.g., the string
            just before the final character is added). The final result
            string itself also takes O(N) space.
    """
    if not isinstance(s, str):
        raise TypeError("Input must be a string.")

    # Base case for the public API: empty or single-character strings are already reversed.
    if len(s) <= 1:
        return s

    # Helper function for the actual recursive logic, using indices to avoid string slicing.
    def _reverse_string_recursive_helper(original_string: str, start_index: int, end_index: int) -> str:
        """
        Recursive helper to reverse a substring defined by start_index and end_index.
        """
        # Base case: If the segment is empty (start_index > end_index), return an empty string.
        # This handles the termination of the recursion when all characters have been processed.
        if start_index > end_index:
            return ""

        # Recursive step:
        # Take the character at the current `end_index` (which is the last character
        # of the current segment) and prepend it to the result of recursively
        # reversing the rest of the string (from `start_index` to `end_index - 1`).
        return original_string[end_index] + _reverse_string_recursive_helper(original_string, start_index, end_index - 1)

    # Initiate the recursion with the full string, from index 0 to len(s) - 1.
    return _reverse_string_recursive_helper(s, 0, len(s) - 1)
