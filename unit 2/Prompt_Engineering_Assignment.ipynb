```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from dotenv import load_dotenv
load_dotenv()

import getpass
import os

if "GOOGLE_API_KEY" not in os.environ:
    os.environ["GOOGLE_API_KEY"] = getpass.getpass("Enter your Google API Key: ")

# 1. Setup (Using the model name that worked for you)
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash", temperature=0)

# 2. The Structured Prompt Template (CO-STAR Framework)
# We use the framework to eliminate ambiguity
prompt_template = ChatPromptTemplate.from_template("""
# Context
You are a Senior Python Developer with 10+ years of experience.

# Objective
Write a Python function to reverse a string.

# Constraints
1. It MUST use recursion.
2. Do NOT use slicing like [::-1].
3. Include detailed Google-style docstrings.

# Style
Professional, clean, and PEP 8 compliant.

# Task
Generate the code for the string: {user_input}
""")

# 3. Create the Chain (LCEL Style)
chain = prompt_template | llm | StrOutputParser()

# 4. Run it
result = chain.invoke({"user_input": "reverse this string"})
print(result)
```
