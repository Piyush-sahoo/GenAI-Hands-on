```python
from dotenv import load_dotenv
load_dotenv()

import getpass
import os
from datetime import datetime
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

# 1. Setup API Key
if "GOOGLE_API_KEY" not in os.environ:
    os.environ["GOOGLE_API_KEY"] = getpass.getpass("Enter your Google API Key: ")

# 2. Initialize using your working model version
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash", temperature=0)

# 3. Define the Template
template = ChatPromptTemplate.from_template(
    """The current year is 2026.
    For the movie '{movie_name}':
    1. Identify its original release year.
    2. Subtract that year from 2026.

    Format: '[Movie Name] was released in [Year], which was [X] years ago.'"""
)

# 4. Create the Chain (LCEL One-Liner)
chain = template | llm | StrOutputParser()

# 5. Run it!
print(chain.invoke({"movie_name": "Inception"}))
```
