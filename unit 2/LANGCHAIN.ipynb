```python
from dotenv import load_dotenv
import os
import getpass

load_dotenv()

if "GOOGLE_API_KEY" not in os.environ:
    os.environ["GOOGLE_API_KEY"] = getpass.getpass("Enter your Google API Key: ")
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI


llm= ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    temperature=0.0
)
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

# Step 1: Create template
template = ChatPromptTemplate.from_template(
    "The movie '{movie}' was released in what year? "
    "How many years ago was that from 2026? "
    "Respond in this format: Year: XXXX | Years Ago: XX"
)

# Step 2: Create parser
parser = StrOutputParser()

# Step 3: Format the prompt
prompt_value = template.invoke({"movie": "Interstellar"})

# Step 4: Call model
response_obj = llm.invoke(prompt_value)

# Step 5: Parse output
final_text = parser.invoke(response_obj)

print(final_text)
```

#LCEL METHOD - ONE LINE CHAIN


```python
chain = (
    ChatPromptTemplate.from_template(
        "The movie '{movie}' was released in what year? "
        "How many years ago was that from 2026? "
        "Respond in this format: Year: XXXX | Years Ago: XX"
    )
    | llm
    | StrOutputParser()
)

print(chain.invoke({"movie": "Interstellar"}))
```
