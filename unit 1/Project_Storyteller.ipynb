```python
from transformers import pipeline, set_seed
import random
storyteller = pipeline('text-generation', model='gpt2')
```

```python
starting_sentence = "The knight entered the dark cave and saw"

print(f"Prompt: {starting_sentence}...")
print("-" * 80)
story = storyteller(
    starting_sentence, 
    max_length=100, 
    do_sample=True, 
    temperature=0.8
)

print(story[0]['generated_text'])
```
