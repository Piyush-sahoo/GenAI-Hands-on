# Unit 1 Mini Project: AI Storyteller

This project generates short creative stories based on a user-provided starting sentence.
It demonstrates text generation using Hugging Face transformer models.


```python
!pip install transformers torch --quiet

from transformers import pipeline
```

```python
prompt = "The knight entered the dark cave"
```

```python
fast_story_gen = pipeline(
    "text-generation",
    model="distilgpt2"
)

fast_output = fast_story_gen(
    prompt,
    max_length=120,
    num_return_sequences=1
)

print("Fast Model Output:\n")
print(fast_output[0]["generated_text"])
```

```python
smart_story_gen = pipeline(
    "text-generation",
    model="gpt2"
)

smart_output = smart_story_gen(
    prompt,
    max_length=120,
    num_return_sequences=1
)

print("Smart Model Output:\n")
print(smart_output[0]["generated_text"])
```

## Observations

- `distilgpt2` generates faster but produces shorter and less detailed stories.
- `gpt2` produces more coherent and creative output but takes longer to generate.
- Both models use decoder-only architectures designed for autoregressive text generation.


```python
new_prompt = "In the year 2050, artificial intelligence became self-aware"

output = smart_story_gen(
    new_prompt,
    max_length=120,
    num_return_sequences=1
)

print(output[0]["generated_text"])
```

## Conclusion

This project demonstrates how transformer-based language models can generate creative text.
By comparing fast and standard models, we observe the trade-off between speed and quality.
